```python
import httpx

url = "https://api.open-meteo.com/v1/forecast"
params = {
    "latitude": 52.52,
    "longitude": 13.41,
    "current": "temperature_2m,relative_humidity_2m,wind_speed_10m",
    "hourly": "temperature_2m,precipitation",
    "daily": "temperature_2m_max,temperature_2m_min,precipitation_sum",
    "timezone": "auto"
}

response = httpx.get(url, params=params)

if response.status_code == 200:
    data = response.json()
    print(data)
else:
    print(f"Error: {response.status_code}")
```

```python
import httpx

response = httpx.get(
    "https://api.open-meteo.com/v1/forecast",
    params={
        "latitude": 55.67594,
        "longitude": 12.56553,
        "current": "temperature_2m,wind_speed_10m",
        "timezone": "America/New_York"
    }
)

data = response.json()

current = data["current"]
print(f"Temperature: {current['temperature_2m']}°C")
print(f"Wind Speed: {current['wind_speed_10m']} km/h")
print(f"Time: {current['time']}")
```

```python
response = httpx.get(
    "https://air-quality-api.open-meteo.com/v1/air-quality",
    params={
        "latitude": 55.6761,
        "longitude": 12.5683,
        "hourly": ["pm10", "pm2_5", "dust", "uv_index"],
	"forecast_days": 1,
    }
)

data = response.json()
data
```

```python
min_pm10 = min(data["hourly"]["pm10"])
max_pm10 = max(data["hourly"]["pm10"])
min_pm2_5 = min(data["hourly"]["pm2_5"])
max_pm2_5 = max(data["hourly"]["pm2_5"])
min_dust = min(data["hourly"]["dust"])
max_dust = max(data["hourly"]["dust"])
min_uv_index = min(data["hourly"]["uv_index"])
max_uv_index = max(data["hourly"]["uv_index"])

print(f"Minimum PM10: {min_pm10}")
print(f"Maximum PM10: {max_pm10}")
print(f"Minimum PM2.5: {min_pm2_5}")
print(f"Maximum PM2.5: {max_pm2_5}")
print(f"Minimum Dust: {min_dust}")
print(f"Maximum Dust: {max_dust}")
print(f"Minimum UV Index: {min_uv_index}")
print(f"Maximum UV Index: {max_uv_index}")
```
